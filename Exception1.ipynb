4.Why LookupError class is used? Explain with an example KeyError and IndexError

Example 1:keyError

```python
try:
  student = {
      "name":"nithiya",
      "age":18

  }
  print(student["mark"])
except KeyError:
  print("not found key")
```

Example 2:IndexError

```python
try:
  a = ["apple","orange","banana"]
  print(a[5])


except IndexError:
  print("not found index")
```

6.List down some best practices for exception handling in python

Example 1:try-except:

```python
try:
  a = 10
  b = 0
  c = a/ b
  print(c)


except ZeroDivisionError:
  print("not divisivble by zero")
```

Example 2:finally use:

```python
try:
    f = open("finall.txt")
    print("File opened successfully")

except FileNotFoundError:
    print("File not found")

finally:
    print("Program finished")
```
