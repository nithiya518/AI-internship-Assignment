1.Write a Python function to calculate the Probability Density Function (PDF) of a normal distribution with mean = 50 and standard deviation = 10 at x = 65. Also plot the complete bell curve and highlight the point x = 65 on the graph.

```python
import scipy.stats as stats
def  calculate_pdf(x,mean,std_dev):
  ab = stats.norm.pdf(x,mean, std_dev)
  return ab

mean = 50
std_dev = 10
x = 65


pdf_at_point = calculate_pdf(x, mean , std_dev)
print(f"pdf at point {x} = {pdf_at_point}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
data = np.random.normal(50,10,1000)


plt.hist(data, bins = 65)
plt.axvline(x=65, color='red', linestyle='--', linewidth=2, label='x = 65')

plt.legend()
plt.show()
```

2.Write a Python function to calculate the Cumulative Distribution Function (CDF) of a normal distribution with mean = 100 and standard deviation = 15 at x = 120. Also print the probability that a value is less than or equal to 120

```python
import scipy.stats as stats
def  calculate_cdf(x,mean,std_dev):
  ab = stats.norm.cdf(x,mean, std_dev)
  return ab

mean = 100
std_dev = 15
x = 120


cdf_at_point = calculate_cdf(x, mean , std_dev)
print(f"cdf at point {x} = {cdf_at_point}")
print(f"the probability that a value is less than or equal to {x} is {cdf_at_point:.4f}")
```
