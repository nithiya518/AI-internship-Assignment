Check if a string is a palindrome.

```python
a="Madam"
b=a.lower()
c=b[::-1]
if b==c:
  print("palindrome")
else:
  print("not palindrome")
```

Find the longest word in a string.


```python
a="welcome to coimbatore and chennai"
words=a.split()
longest=words[0]
for i in words:
  if len(i)>len(longest):
    longest=i


print(longest)
```

Find the shortest word in a string.


```python
a="welcome to coimbatore  tcs"
words=a.split()
shortest=words[3]
for t in words:
  if len(t)<len(shortest):
   shortest=t

print(shortest)
```
