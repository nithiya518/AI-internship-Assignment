```python
username = "nithi@gmail.com"

if username.count("@")==1:
  parts = username.index("@")

  if parts > 0 and  parts < len(username)-1:
    print("valid email")
  else:
    print("not valid")

else:
  print("not valid")

```

27.Convert a CSV string to a list of lists.

```python
a = "apple,10,red\norange,10,orange\nbanana,10,yellow"
b = a.split("\n")
result =[]
for i in b:
  items = i.split(",")
  result.append(items)
print(result)
```
